```python
img = cv2.imread("./images/arpitImg.JPG")
img = cv2.cvtColor(img , cv2.COLOR_BGR2RGB)
plt.imshow(img)
plt.axis("off")
plt.show()
```

```python
image1 = img[:,:,0]
image2 = img[:,:,1]
image3 = img[:,:,2]
combined_Image = np.hstack((image1,image2,image3))
print(image1.shape)
print(image2.shape)
print(image3.shape)
plt.imshow(combined_Image)
plt.axis("off")
plt.show()
```

```python
img_sz = cv2.resize(img , (5216,5216))
plt.imshow(img_sz)
plt.axis("off")
plt.show()
```

```python
img_flip_topBot = cv2.flip(img , 0)
img_flip_rgtLft = cv2.flip(img , 1)
img_flip_bothRev = cv2.flip(img , -1)

allImg = np.hstack((img_flip_topBot,img_flip_rgtLft,img_flip_bothRev))
plt.imshow(allImg)
plt.axis("off")
plt.show()
```

```python
img_cropped = img[1000:3000, 1000:3000]

img_cropped_rgb = cv2.cvtColor(img_cropped, cv2.COLOR_RGB2BGR)

plt.imshow(img_cropped_rgb)
plt.axis("off")
plt.show()
cv2.imwrite('cropped_img.jpg',img_cropped_rgb)
```

```python
img_create = np.zeros((512,512,3))
plt.imshow(img_create)
plt.axis("off")
plt.show()
```

```python
cv2.rectangle(img_create , pt1=(100,100),pt2=(300,300),color=(255,0,0),thickness=3)
plt.imshow(img_create)
plt.axis("off")
plt.show()
```

```python
cv2.circle(img_create , center=(200,200) , radius=40 , color=(0,255,0) , thickness=-1)
plt.imshow(img_create)
plt.axis("off")
plt.show()
```
