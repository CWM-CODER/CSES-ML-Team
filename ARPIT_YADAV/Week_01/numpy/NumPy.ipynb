## 2. Why NumPy?

A Python list can store numbers, but NumPy arrays are designed specifically for numerical computation.

### Python list
- General-purpose container
- Can contain mixed data types
- Arithmetic usually requires explicit loops

### NumPy ndarray
- Homogeneous numerical data is stored efficiently
- Supports vectorized operations
- Supports multidimensional arrays
- Provides fast mathematical, statistical and linear-algebra operations

The central object in NumPy is the **ndarray** (N-dimensional array).

````python
# Python list
marks = [70, 80, 90, 60]

# NumPy array
marks_np = np.array(marks)

print("Python list :", marks)
print("NumPy array :", marks_np)

# Vectorized operation: add 5 marks to every student
print("Updated marks:", marks_np + 5)
````

## 3. Creating NumPy arrays

- `ndim` → number of dimensions
- `shape` → size along each dimension

- `dtype` → data type of elements

## 10. Vectorized arithmetic

````python
a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30, 40])

print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)
print("a ** 2 =", a ** 2)

# Scalar broadcasting
print("a + 100 =", a + 100)
print("a * 10  =", a * 10)
````

### Element-wise operations vs matrix operations

This distinction is extremely important.

- `A * B` → element-wise multiplication
- `A @ B` → matrix multiplication
- `np.matmul(A, B)` → matrix multiplication
- `np.dot(A, B)` → dot product / matrix multiplication depending on dimensions

Do not confuse `*` with matrix multiplication.

````python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("A * B (element-wise):\n", A * B)
print("\nA @ B (matrix multiplication):\n", A @ B)
````

## 11. Comparison operators and Boolean arrays

````python
x = np.array([10, 20, 30, 40, 50])

print(x > 25)
print(x == 30)
print(x != 30)
print(x <= 30)

# Combine conditions with &, | and ~
print("\nBetween 20 and 40:")
print((x >= 20) & (x <= 40))

print("\nLess than 20 OR greater than 40:")
print((x < 20) | (x > 40))
````

**Important:** With NumPy arrays, use `&`, `|`, and `~` for element-wise logical operations.

Use parentheses around each condition:

`(x > 10) & (x < 50)`

Do not write:

`x > 10 and x < 50`

## 12. Boolean masking / filtering

````python
marks = np.array([35, 78, 92, 41, 66, 29, 88])

passed = marks >= 40

print("Marks :", marks)
print("Mask  :", passed)
print("Passed:", marks[passed])

# Directly use the condition
print("Marks >= 70:", marks[marks >= 70])

# Multiple conditions
print("40 <= marks < 80:", marks[(marks >= 40) & (marks < 80)])
````

## 13. np.where()

````python
marks = np.array([35, 78, 92, 41, 66, 29, 88])

# np.where(condition, value_if_true, value_if_false)
result = np.where(marks >= 40, "Pass", "Fail")
print(result)

# Add grace marks to students below 40
adjusted = np.where(marks < 40, marks + 5, marks)
print("Adjusted marks:", adjusted)
````

## 16. Axis: one of the most important concepts

For a 2-D array:

- `axis=0` → operate **down the rows**, producing one result per column
- `axis=1` → operate **across the columns**, producing one result per row

Think carefully about the dimension that is being **collapsed**.

````python
marks = np.array([
    [80, 70, 90],
    [60, 75, 85],
    [90, 95, 88]
])

print("Marks:\n", marks)

print("\nColumn-wise mean (axis=0):", np.mean(marks, axis=0))
print("Row-wise mean    (axis=1):", np.mean(marks, axis=1))

print("\nColumn-wise maximum:", np.max(marks, axis=0))
print("Row-wise maximum   :", np.max(marks, axis=1))

print("\nColumn-wise sum:", np.sum(marks, axis=0))
print("Row-wise sum   :", np.sum(marks, axis=1))
````

## 18. Reshaping arrays

````python
x = np.arange(1, 13)
print("Original:", x)
print("Shape:", x.shape)

A = x.reshape(3, 4)
print("\n3 x 4:\n", A)

B = x.reshape(4, 3)
print("\n4 x 3:\n", B)

# One dimension can be inferred using -1
C = x.reshape(3, -1)
print("\n3 x inferred:\n", C)
````

### Reshape rule

The total number of elements must remain unchanged.

For example:

12 elements → valid shapes include `(3, 4)`, `(4, 3)`, `(2, 6)`, `(12,)`

But `(5, 3)` is invalid because it requires 15 elements.

````python
try:
    x.reshape(5, 3)
except ValueError as e:
    print("Error:", e)
````

## 23. Broadcasting — essential for ML

````python
# A scalar broadcasts to every element
x = np.array([1, 2, 3])
print(x + 10)

# A row vector can be added to every row
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

bias = np.array([10, 20, 30])
print("\nA + bias:\n", A + bias)
````

### Broadcasting intuition

Suppose:

- `A.shape == (3, 4)`
- `b.shape == (4,)`

NumPy treats `b` as compatible with each row of `A`.

Another common example:

- `A.shape == (3, 1)`
- `b.shape == (1, 4)`

The result can have shape `(3, 4)`.

Broadcasting is heavily used in machine learning for normalization, biases, feature scaling and batch operations.

````python
A = np.arange(1, 4).reshape(3, 1)
B = np.arange(10, 14).reshape(1, 4)

print("A shape:", A.shape)
print("B shape:", B.shape)

C = A + B
print("\nA + B:\n", C)
print("Result shape:", C.shape)
````

## 33. Matrix inverse, determinant and eigenvalues

````python
A = np.array([
    [4, 2],
    [1, 3]
], dtype=float)

print("Determinant:", np.linalg.det(A))
print("\nInverse:\n", np.linalg.inv(A))

eigenvalues, eigenvectors = np.linalg.eig(A)

print("\nEigenvalues:", eigenvalues)
print("\nEigenvectors:\n", eigenvectors)
````

### Practical advice

Although `np.linalg.inv(A)` is useful for learning, **do not normally compute an inverse just to solve `Ax=b`**.

Prefer:

`np.linalg.solve(A, b)`

It is generally more numerically appropriate and efficient.

## 38. Views vs copies — very important

````python
original = np.array([10, 20, 30, 40, 50])

# Slicing generally creates a view
view = original[1:4]
view[0] = 999

print("Original after modifying view:", original)

# Explicit copy
original2 = np.array([10, 20, 30, 40, 50])
copy = original2[1:4].copy()
copy[0] = 999

print("Original after modifying copy:", original2)
print("Copy:", copy)
````

### Rule of thumb

If you need an independent array:

`new_array = old_array.copy()`

This avoids accidental modification of the original data.

## 39. Copying an array vs assigning a variable

````python
a = np.array([1, 2, 3])

# Assignment creates another reference to the same array
b = a
b[0] = 100

print("a after b[0] change:", a)

# copy() creates independent data
c = np.array([1, 2, 3])
d = c.copy()
d[0] = 100

print("c after d[0] change:", c)
print("d:", d)
````

## 44. Performance: Python loop vs vectorization

````python
import time

n = 1_000_000
x = np.arange(n, dtype=np.float64)

# Python list computation
python_list = list(range(n))

start = time.perf_counter()
result_list = [v * 2 + 1 for v in python_list]
list_time = time.perf_counter() - start

# NumPy vectorized computation
start = time.perf_counter()
result_numpy = x * 2 + 1
numpy_time = time.perf_counter() - start

print(f"Python list comprehension: {list_time:.6f} seconds")
print(f"NumPy vectorization     : {numpy_time:.6f} seconds")
print(f"Speed ratio             : {list_time / numpy_time:.2f}x")
````

### Main lesson

Instead of:

```python
for i in range(len(x)):
    y[i] = x[i] * 2 + 1
```

prefer:

```python
y = x * 2 + 1
```

This is called **vectorization**.

NumPy pushes many operations into optimized compiled implementations, reducing Python-level loop overhead.

## 45. Vectorized normalization — common ML operation

````python
# Suppose each row represents one student's marks in three subjects.
X = np.array([
    [80, 70, 90],
    [60, 75, 85],
    [90, 95, 88]
], dtype=float)

# Standardization column-wise:
# z = (x - mean) / standard deviation
mean = X.mean(axis=0)
std = X.std(axis=0)

X_standardized = (X - mean) / std

print("Mean:", mean)
print("Std :", std)
print("\nStandardized data:\n", X_standardized)

print("\nNew column means:", X_standardized.mean(axis=0))
print("New column stds :", X_standardized.std(axis=0))
````

## 46. Min-Max normalization

````python
X = np.array([
    [10, 100],
    [20, 200],
    [30, 300],
    [40, 400]
], dtype=float)

X_min = X.min(axis=0)
X_max = X.max(axis=0)

X_scaled = (X - X_min) / (X_max - X_min)

print("Original:\n", X)
print("\nMin:", X_min)
print("Max:", X_max)
print("\nScaled:\n", X_scaled)
````

## 47. Pairwise distance example

````python
# Three points in 2-D
points = np.array([
    [1, 2],
    [4, 6],
    [7, 3]
], dtype=float)

# Difference between every pair of points using broadcasting
diff = points[:, np.newaxis, :] - points[np.newaxis, :, :]

# Euclidean distance
distances = np.sqrt(np.sum(diff**2, axis=2))

print("Points:\n", points)
print("\nPairwise distance matrix:\n", distances)
````

This pattern is useful in:

- K-Nearest Neighbors
- clustering
- recommender systems
- graph construction
- similarity search

The important idea is to use broadcasting rather than writing nested Python loops.

## 48. Cosine similarity

````python
def cosine_similarity(a, b):
    # Convert inputs to NumPy arrays for consistent behavior.
    a = np.asarray(a, dtype=float)
    b = np.asarray(b, dtype=float)

    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

v1 = np.array([1, 2, 3])
v2 = np.array([2, 4, 6])
v3 = np.array([3, 0, 1])

print("cos(v1, v2):", cosine_similarity(v1, v2))
print("cos(v1, v3):", cosine_similarity(v1, v3))
````

## 59. A small NumPy-based linear regression calculation

````python
# Generate synthetic y = Xw + noise
rng = np.random.default_rng(10)

X = rng.normal(size=(200, 2))
true_w = np.array([3.0, -2.0])
true_b = 5.0
noise = rng.normal(scale=0.5, size=200)

y = X @ true_w + true_b + noise

# Add a column of ones to learn the intercept together with weights.
X_design = np.column_stack([np.ones(len(X)), X])

# Least-squares solution
theta, residuals, rank, singular_values = np.linalg.lstsq(
    X_design, y, rcond=None
)

print("Estimated intercept:", theta[0])
print("Estimated weights  :", theta[1:])
print("True intercept     :", true_b)
print("True weights       :", true_w)
````

## 60. Common NumPy mistakes

````python
# Mistake 1: using * when you mean matrix multiplication
A = np.eye(2)
B = np.ones((2, 2))

print("A * B = element-wise multiplication")
print(A * B)

print("\nA @ B = matrix multiplication")
print(A @ B)
````

### Common mistakes to discuss with students

1. **Confusing shape and size**
   - `(3, 4)` is a shape.
   - `12` is the size.

2. **Confusing `*` and `@`**
   - `*` is element-wise.
   - `@` is matrix multiplication.

3. **Forgetting `axis`**
   - Always ask: "Which dimension am I collapsing?"

4. **Using Python `and` / `or`**
   - Use `&` / `|` for element-wise array conditions.

5. **Accidental modification through a view**
   - Use `.copy()` when an independent array is required.

6. **Ignoring dtype**
   - Integer arrays cannot store fractional values correctly.

7. **Using `np.linalg.inv()` unnecessarily**
   - Prefer `np.linalg.solve()` for solving linear systems.

8. **Writing slow Python loops**
   - Look for a vectorized NumPy operation.

9. **Ignoring broadcasting rules**
   - Check the shapes before performing operations.

10. **Assuming `np.empty()` initializes to zero**
    - It does not.
