```python
# view vs copy
org = np.array([1,2,3,4])
tmp = org # given by the reference
hlp = org.copy() # now only copy on changing the hlp do not chnage the org
x = org.view() # copied by the reference
tmp[0]=-1
hlp[0]=10
x[2]=88
print(org)
```

```python
#unique values and the frequency
arr = np.array([1,2,3,4,5,1,1,2,3,3,5,2,2,2,23,4])
unique = np.unique(arr)
print(unique)
unique , cnt = np.unique(arr , return_counts=True)
x  = dict(zip(unique,cnt))
print(x)
```

```python
tmp = np.where(arr>4,-1,arr)
print(tmp)
```
