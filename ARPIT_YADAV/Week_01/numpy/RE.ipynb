# Regular Expressions

A **Regular Expression (Regex)** is a sequence of characters used to define a **search pattern**.  
It is commonly used to **find, match, extract, validate, or replace text**.

For example:
- `cat` → matches the exact word **cat**
- `\d` → matches any digit
- `\w+` → matches one or more word characters

## Common Regular Expression Symbols

| Symbol | Meaning | Example | Matches |
|---|---|---|---|
| `.` | Any single character | `c.t` | `cat`, `cut`, `cot` |
| `^` | Start of a string | `^Hello` | `Hello world` |
| `$` | End of a string | `world$` | `Hello world` |
| `*` | Zero or more occurrences | `ab*` | `a`, `ab`, `abb` |
| `+` | One or more occurrences | `ab+` | `ab`, `abb` |
| `?` | Zero or one occurrence | `colou?r` | `color`, `colour` |
| `{n}` | Exactly `n` occurrences | `\d{3}` | `123`, `456` |
| `{n,}` | At least `n` occurrences | `\d{3,}` | `123`, `1234` |
| `{n,m}` | Between `n` and `m` occurrences | `\d{2,4}` | `12`, `123`, `1234` |
| `[]` | Character set | `[aeiou]` | `a`, `e`, `i`, `o`, `u` |
| `[^]` | Negated character set | `[^0-9]` | Any non-digit |
| `|` | OR / alternative | `cat|dog` | `cat` or `dog` |
| `()` | Grouping | `(ab)+` | `ab`, `abab` |
| `\d` | Any digit (`0–9`) | `\d+` | `123`, `45` |
| `\D` | Any non-digit | `\D+` | `abc`, `hello` |
| `\w` | Word character | `\w+` | `hello`, `abc123` |
| `\W` | Non-word character | `\W+` | `@`, `#`, `!` |
| `\s` | Whitespace | `\s+` | Space, tab, newline |
| `\S` | Non-whitespace | `\S+` | `hello`, `123` |
| `\b` | Word boundary | `\bcat\b` | `cat` but not `catch` |
| `\\` | Literal backslash | `\\` | `\` |

## Simple Examples

| Regular Expression | Purpose | Example Match |
|---|---|---|
| `\d+` | Find one or more digits | `12345` |
| `[A-Z]` | Find an uppercase letter | `A` |
| `[a-z]+` | Find lowercase words | `hello` |
| `\w+@\w+\.\w+` | Simple email pattern | `user@gmail.com` |
| `^Hello` | Text starts with "Hello" | `Hello World` |
| `World$` | Text ends with "World" | `Hello World` |

> **Key idea:** Regular expressions combine ordinary characters and special symbols to describe **patterns in text**.

```python
import re

# Exact String Matching
text = "I am learning Python"

result = re.search("Python", text)

print(result.group(0))
```

```python
# Dot (.) - Any Single Character
text = "catsa cot cut cit bat"

matches = re.findall(r"c.t", text)

print(matches)
```

```python
# Caret (^) - Start of String
text = "Python is easy"

result = re.search(r"^Python", text)

print(result)
```

```python
# Dollar ($) - End of String
text = "I am learning Python"

result = re.search(r"Python$", text)

print(result)
```

```python
# Character Set ([])
text = "cat bat rat mat zat"

matches = re.findall(r"[cbrm]at", text)

print(matches)
```

```python
# Character Range
text = "apple Banana CAT 123"

lowercase = re.findall(r"[a-z]+", text)
uppercase = re.findall(r"[A-Z]+", text)
digits = re.findall(r"[0-9]+", text)

print("Lowercase:", lowercase)
print("Uppercase:", uppercase)
print("Digits:", digits)
```

```python
# Negated Character Set ([^])
text = "abc123XYZ"

matches = re.findall(r"[^0-9]+", text)

print(matches)
```

```python
# \d - Any Digit
text = "My marks are 85 and 92"

matches = re.findall(r"\d+", text)

print(matches)
```

```python
# \D - Any Non-Digit
text = "abc123XYZ"

matches = re.findall(r"\D+", text)

print(matches)
```

```python
# \w - Word Character
text = "hello world_123 @python"

matches = re.findall(r"\w+", text)

print(matches)
```

```python
# \W - Non-Word Character
text = "hello@world.com!"

matches = re.findall(r"\W+", text)

print(matches)
```

```python
# \s - Whitespace
text = "Hello   World\tPython"

matches = re.findall(r"\s+", text)

print(matches)
```

```python
# \S - Non-Whitespace
text = "Hello World Python"

matches = re.findall(r"\S+", text)

print(matches)
```

```python
# * - Zero or More Occurrences
text = "a ab abb abbb abab"

matches = re.findall(r"ab*", text)

print(matches)
```

```python
# + - One or More Occurrences
text = "a ab abab abbb"

matches = re.findall(r"(ab)+", text)

print(matches)
```

```python
# ? - Zero or One Occurrence
text = "color colour colouur"

matches = re.findall(r"colou?r", text)

print(matches)
```

```python
# {n} - Exactly n Occurrences
text = "1 12  123 1234 12345"

matches = re.findall(r"\d{3}", text)

print(matches)
```

```python
# {n,} - At Least n Occurrences
text = "1 12 123 1234 12345"

matches = re.findall(r"\d{3,}", text)

print(matches)
```

```python
# {n,m} - Between n and m Occurrences
text = "1 12 123 1234 12345"

matches = re.findall(r"\d{2,4}", text)

print(matches)
```

```python
# | - OR Operator
text = "I like cats and dogs"

matches = re.findall(r"cats|dogs", text)

print(matches)
```

```python
# Without grouping

text = "cat bat rat"

matches = re.findall(r"cat|bat", text)

print(matches)
```

```python
# With grouping

text = "cat bat rat"

matches = re.findall(r"(cat|bat)", text)

print(matches)
```

```python
# () - Grouping
text = "ha ha haha hahaha"

matches = re.findall(r"(?:ha)+", text)

print(matches)
```

```python
# Capturing Groups
text = "Name: Ravi, Age: 25"

match = re.search(r"Name: (\w+), Age: (\d+)", text)

print("Name:", match.group(1))
print("Age:", match.group(2))
```

```python
# Named Groups
text = "Name: Ravi, Age: 25"

match = re.search(
    r"Name: (?P<name>\w+), Age: (?P<age>\d+)",
    text
)

print("Name:", match.group("name"))
print("Age:", match.group("age"))
```

```python
# \b - Word Boundary
text = "cat category bobcat cat"

matches = re.findall(r"\bcat\b", text)

print(matches)
```

```python
# re.search() - Find First Match
text = "Python is easy. Python is popular."

match = re.search(r"Python", text)

print(match.group())
print("Start:", match.start())
print("End:", match.end())
```

```python
# re.match() - Match from Beginning
text = "Python is easy"

match = re.match(r"Python", text)

print(match)
```

```python
# re.fullmatch() - Match the Entire String
text = "12345a"

match = re.fullmatch(r"\d+", text)

if match:
    print("The entire string contains only digits")
else:
    print("Not a valid number")
```

```python
# re.findall() - Find All Matches
text = "Python, Java, Python, C++, Python"

matches = re.findall(r"Python", text)

print(matches)
print("Number of matches:", len(matches))
```

```python
# re.finditer() - Iterate Through All Matches
text = "Python is easy. Python is popular."

for match in re.finditer(r"is", text):
    print("Match:", match.group())
    print("Start:", match.start())
    print("End:", match.end())
```

```python
# re.split() - Split Using a Regular Expression
text = "apple,banana;orange|grape"

items = re.split(r"[,;|]", text)

print(items)
```

```python
# re.sub() - Replace Matching Text
text = "My phone number is 9876543210"

result = re.sub(r"\d", "X", text)

print(result)
```

```python
# Replacing Multiple Spaces
text = "Python    is   very     useful"

result = re.sub(r"\s+", " ", text)

print(result)
```

```python
# Case-Insensitive Matching
text = "Python PYTHON python PyThOn"

matches = re.findall(r"python", text, re.IGNORECASE)

print(matches)
```

```python
# Multiline Matching
text = """Python is easy
Java is popular
Python is powerful"""

matches = re.findall(r"^Python", text, re.MULTILINE)

print(matches)
```

```python
# Raw Strings
text = "There are 25 students"

pattern = r"\d+"

matches = re.findall(pattern, text)

print(matches)
```

```python
# Extract Email Addresses
text = """
Contact us at alice@gmail.com or bob@yahoo.com
"""

pattern = r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}"

emails = re.findall(pattern, text)

print(emails)
```

```python
# Extract Phone Numbers
text = """
Call me at 9876543210 or 9123456789.
"""

pattern = r"\b\d{10}\b"

phones = re.findall(pattern, text)

print(phones)
```

```python
# Extract Dates
text = "Important dates are 12-08-2026 and 25/12/2026."

pattern = r"\b\d{2}[-/]\d{2}[-/]\d{4}\b"

dates = re.findall(pattern, text)

print(dates)
```

```python
# Validate an Email Address
email = "student@example.com"

pattern = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"

if re.fullmatch(pattern, email):
    print("Valid email")
else:
    print("Invalid email")
```

```python
# Validate a 10-Digit Indian Mobile Number
phone = "9876543210"

pattern = r"^[6-9]\d{9}$"

if re.fullmatch(pattern, phone):
    print("Valid mobile number")
else:
    print("Invalid mobile number")
```

```python
# Extract Hashtags
text = "Learning #Python and #MachineLearning is fun!"

hashtags = re.findall(r"#\w+", text)

print(hashtags)
```

```python
# Extract Mentions
text = "Hello @ravi and @student123"

mentions = re.findall(r"@\w+", text)

print(mentions)
```

```python
# Extract URLs
text = """
Visit https://www.google.com or https://www.python.org
"""

pattern = r"https?://[^\s]+"

urls = re.findall(pattern, text)

print(urls)
```

```python
# Positive Lookahead (?=...)
# Find numbers followed by "kg"
text = "10kg 20m 30kg"

matches = re.findall(r"\d+(?=kg)", text)

print(matches)
```

```python
# Negative Lookahead (?!...)
# Find numbers NOT followed by "kg"
text = "10kg 20m 30kg"

matches = re.findall(r"\d+(?!kg)", text)

print(matches)
```

```python
# Positive Lookbehind (?<=...)
# Find numbers preceded by the Rupee symbol
text = "Price ₹500 and ₹1000"

matches = re.findall(r"(?<=₹)\d+", text)

print(matches)
```

```python
# Negative Lookbehind (?<!...)
# Find numbers NOT preceded by the Rupee symbol
text = "₹500 600 ₹700 800"

matches = re.findall(r"(?<!₹)\d+", text)

print(matches)
```

```python
import re

# Practical Example - Extract Information from Text

text = """
Name: Ravi
Email: ravi@gmail.com
Phone: 9876543210
Age: 25
"""

# Extract Name
name = re.search(r"Name:\s*(\w+)", text).group(1)

# Extract Email
email = re.search(r"[\w.-]+@[\w.-]+\.\w+", text).group()

# Extract Phone
phone = re.search(r"\b[6-9]\d{9}\b", text).group()

# Extract Age
age = re.search(r"Age:\s*(\d+)", text).group(1)

print("Name :", name)
print("Email:", email)
print("Phone:", phone)
print("Age  :", age)
```

### Regular Expression Cheat Sheet

| Pattern | Meaning |
|---|---|
| `.` | Any character |
| `^` | Start of string |
| `$` | End of string |
| `*` | Zero or more |
| `+` | One or more |
| `?` | Zero or one |
| `{n}` | Exactly n |
| `{n,}` | At least n |
| `{n,m}` | Between n and m |
| `[]` | Character set |
| `[^]` | Negated character set |
| `|` | OR |
| `()` | Group |
| `\d` | Digit |
| `\D` | Non-digit |
| `\w` | Word character |
| `\W` | Non-word character |
| `\s` | Whitespace |
| `\S` | Non-whitespace |
| `\b` | Word boundary |
| `(?=...)` | Positive lookahead |
| `(?!...)` | Negative lookahead |
| `(?<=...)` | Positive lookbehind |
| `(?<!...)` | Negative lookbehind |

### Important Python Functions

| Function | Purpose |
|---|---|
| `re.search()` | Find the first match anywhere |
| `re.match()` | Match from the beginning |
| `re.fullmatch()` | Match the entire string |
| `re.findall()` | Find all matches |
| `re.finditer()` | Iterate over all matches |
| `re.split()` | Split using a pattern |
| `re.sub()` | Replace matched text |
| `re.compile()` | Create a reusable regex pattern |
