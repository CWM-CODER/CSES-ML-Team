```python
df = pd.read_csv('sharma-kohli.csv')
df
```

```python
plt.plot(df['index'] , df['V Kohli'])
plt.plot(df['index'] , df['RG Sharma'])
```

```python
plt.plot(df['index'] , df['V Kohli'])
plt.plot(df['index'] , df['RG Sharma'])
plt.title("Rohit Sharma Vs Virat Kohli Career Comparison")
plt.xlabel("Season")
plt.ylabel("Runs Scored")
```

```python
plt.plot(df['index'] , df['V Kohli'] , color="green")
plt.plot(df['index'] , df['RG Sharma'] , color="red")
plt.title("Rohit Sharma Vs Virat Kohli Career Comparison")
plt.xlabel("Season")
plt.ylabel("Runs Scored")
```

```python
plt.plot(df['index'] , df['V Kohli'] , linestyle="dashed")
plt.plot(df['index'] , df['RG Sharma'],linestyle="dotted")
plt.title("Rohit Sharma Vs Virat Kohli Career Comparison")
plt.xlabel("Season")
plt.ylabel("Runs Scored")
```

```python
plt.plot(df['index'] , df['V Kohli'] , linestyle="dashdot")
plt.plot(df['index'] , df['RG Sharma'],linestyle="dashdot")
plt.title("Rohit Sharma Vs Virat Kohli Career Comparison")
plt.xlabel("Season")
plt.ylabel("Runs Scored")
```

```python
plt.plot(df['index'] , df['V Kohli'] , linewidth="3")
plt.plot(df['index'] , df['RG Sharma'],linewidth="5")
plt.title("Rohit Sharma Vs Virat Kohli Career Comparison")
plt.xlabel("Season")
plt.ylabel("Runs Scored")
```

```python
plt.plot(df['index'] , df['V Kohli']  , marker=".")
plt.plot(df['index'] , df['RG Sharma'],marker="d")
plt.title("Rohit Sharma Vs Virat Kohli Career Comparison")
plt.xlabel("Season")
plt.ylabel("Runs Scored")
```

```python
plt.plot(df['index'] , df['V Kohli']  , marker="." , markersize="30")
plt.plot(df['index'] , df['RG Sharma'],marker="d")
plt.title("Rohit Sharma Vs Virat Kohli Career Comparison")
plt.xlabel("Season")
plt.ylabel("Runs Scored")
```

```python
plt.plot(df['index'] , df['V Kohli']  , marker=".",label="virat")
plt.plot(df['index'] , df['RG Sharma'],marker="d",label="rohit")
plt.title("Rohit Sharma Vs Virat Kohli Career Comparison")
plt.xlabel("Season")
plt.ylabel("Runs Scored")
plt.legend() # this is for individual line labeling 
```
